# 사극체 말투 변환기 — QLoRA 4bit

이 노트북은 기존 `test.py`를 기준으로 다음 문제를 막도록 정리한 개선판입니다.

1. **LoRA adapter가 붙었지만 trainable이 0개가 되는 문제**를 `ensure_lora_trainable()`로 강제 점검합니다.
2. `prompt` / `completion` 컬럼만 `SFTTrainer`에 넘겨서 데이터 포맷 해석을 명확하게 합니다.
3. `completion_only_loss=True`가 실제로 completion 토큰에만 걸리는지 확인하는 진단 셀을 넣었습니다.
4. 작은 데이터셋에서 loss가 출렁이는 문제를 고려해 `max_steps`, `logging_steps`, `warmup_ratio`, `max_grad_norm`을 안정적으로 조정했습니다.
5. Qwen 계열 chat template의 EOS 토큰이 `<|im_end|>`인지 확인하고 `SFTConfig(eos_token=...)`에 명시합니다.

> 권장 실행 순서: **커널 재시작 → 위에서부터 순서대로 Run All**.  
> LoRA 설정 셀을 중복 실행하면 adapter가 중첩될 수 있으므로, 중간부터 다시 돌릴 때는 모델 로드 셀부터 다시 실행하세요.


QLoRA/LoRA 파인튜닝에서 자주 쓰는 라이브러리들이네요. 각각의 역할입니다.

transformers (Hugging Face): 사전학습된 모델(예: LLaMA, OPT 등)과 토크나이저를 불러오고, 학습/추론 파이프라인을 제공하는 핵심 라이브러리입니다. 모델 아키텍처 정의, 가중치 로딩, Trainer 등을 담당합니다.

peft (Parameter-Efficient Fine-Tuning): LoRA, QLoRA 같은 파라미터 효율적 파인튜닝 기법을 구현한 라이브러리입니다. 전체 모델을 다 학습하지 않고, 소수의 어댑터 파라미터만 추가해 학습하도록 모델을 감싸주는(wrapping) 역할을 합니다.

trl (Transformer Reinforcement Learning): SFT(지도 파인튜닝), RLHF, DPO 등 LLM 파인튜닝을 위한 고수준 학습 유틸리티를 제공합니다. SFTTrainer 같은 클래스로 학습 루프를 간편하게 구성할 수 있게 해줍니다.

bitsandbytes: 4bit/8bit 양자화(quantization) 연산을 지원하는 라이브러리입니다. QLoRA에서 모델을 4bit로 로드해 메모리 사용량을 크게 줄이는 데 핵심적인 역할을 합니다.

accelerate: 멀티 GPU/분산 학습, mixed precision, 디바이스 배치 등을 추상화해서 코드 수정 없이 다양한 하드웨어 환경에서 학습을 실행할 수 있게 해주는 라이브러리입니다.

정리하면 transformers(모델)+peft(어댑터 기법)+trl(학습 루프)+bitsandbytes(양자화)+accelerate(분산/디바이스 관리)가 합쳐져 QLoRA 파인튜닝 스택을 구성합니다.

## 2. 모델과 토크나이저 로드 — NF4 4bit QLoRA

`MODEL_ID`만 바꾸면 1.7B 또는 4B 계열 모두 같은 구조로 실험할 수 있습니다. 6GB VRAM에서는 1.7B가 안전하고, 더 큰 GPU에서는 4B도 시도할 수 있습니다.


```python
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
# "./lora_adapters/sample_qlora_qwen1-7b_sageuk"
MODEL_ID = "Qwen/Qwen3-1.7B"  # 예: "Qwen/Qwen3-4B-Instruct-2507"

# 4bit(NF4) 양자화 설정: 모델 가중치를 4bit로 로드해 VRAM 사용량을 크게 줄입니다.
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=COMPUTE_DTYPE,  # 실제 행렬곱 연산은 이 dtype으로 수행됩니다.
    bnb_4bit_use_double_quant=True,  # 양자화 상수까지 한 번 더 양자화해 메모리를 더 아낍니다.
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)

# Qwen chat template의 assistant 종료 토큰은 보통 <|im_end|>입니다.
# SFTConfig에 eos_token으로 명시하면 학습/생성 종료가 더 안정적입니다.
vocab = tokenizer.get_vocab()
EOS_TOKEN = "<|im_end|>" if "<|im_end|>" in vocab else tokenizer.eos_token
if EOS_TOKEN is None:
    raise ValueError("tokenizer.eos_token을 찾지 못했습니다. 모델 토크나이저를 확인하세요.")

tokenizer.eos_token = EOS_TOKEN
if tokenizer.pad_token is None:
    # pad_token이 없으면 배치로 묶을 때 패딩할 토큰이 없어 에러가 나므로, EOS로 대체합니다.
    tokenizer.pad_token = EOS_TOKEN

def load_causal_lm_4bit(model_id: str):
    kwargs = dict(
        quantization_config=bnb_config,
        device_map={"": 0},  # 모델 전체를 GPU 0번에 올립니다(단일 GPU, 분산 없음).
        trust_remote_code=True,
    )
    # transformers 버전에 따라 dtype / torch_dtype 인자명이 다를 수 있어 fallback을 둡니다.
    try:
        return AutoModelForCausalLM.from_pretrained(model_id, dtype=COMPUTE_DTYPE, **kwargs)
    except TypeError:
        return AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=COMPUTE_DTYPE, **kwargs)

model = load_causal_lm_4bit(MODEL_ID)
# 학습 중에는 KV 캐시가 필요 없고 gradient checkpointing과 충돌할 수 있어 꺼둡니다.
model.config.use_cache = False

print("모델 로드 완료")
print("EOS_TOKEN:", repr(EOS_TOKEN), "id=", tokenizer.eos_token_id)
print("PAD_TOKEN:", repr(tokenizer.pad_token), "id=", tokenizer.pad_token_id)
print(f"현재 GPU 메모리 사용량: {torch.cuda.memory_allocated()/1024**3:.2f} GB")
```

## 3. 공통 유틸리티

핵심은 `ensure_lora_trainable()`입니다. 사용자가 겪은 `LoRA params는 있는데 Trainable LoRA params가 0개`인 상태를 여기서 바로 잡거나, 잡히지 않으면 명확히 에러를 냅니다.


```python
def clear_cuda_cache():
    '''GPU 메모리 파편화를 줄이기 위해 가비지 컬렉션 후 CUDA 캐시를 비웁니다.'''
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


def model_device(m):
    '''모델 파라미터가 올라가 있는 device(cpu/cuda)를 반환합니다.'''
    return next(m.parameters()).device


def apply_chat(messages, add_generation_prompt: bool):
    '''Qwen3의 enable_thinking=False를 우선 사용하고, 미지원 모델이면 일반 chat template로 fallback합니다.'''
    try:
        return tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=add_generation_prompt,
            enable_thinking=False,
        )
    except TypeError:
        return tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=add_generation_prompt,
        )


def strip_qwen_think(text: str) -> str:
    '''혹시 <think>...</think>가 섞여 나오면 최종 답변만 남깁니다.'''
    if "</think>" in text:
        text = text.split("</think>", 1)[1]
    return text.replace(EOS_TOKEN, "").strip()


def ask(question: str, max_new_tokens: int = 120, do_sample: bool = False) -> str:
    '''질문 하나를 chat template에 맞춰 모델에 넣고, 생성된 답변 텍스트만 뽑아 반환합니다.'''
    model.eval()
    messages = [{"role": "user", "content": question}]
    prompt_text = apply_chat(messages, add_generation_prompt=True)
    inputs = tokenizer(prompt_text, return_tensors="pt").to(model_device(model))

    # GPU에서는 autocast로 연산 dtype을 맞춰 속도/메모리를 최적화합니다.
    autocast_ctx = torch.autocast(device_type="cuda", dtype=COMPUTE_DTYPE) if torch.cuda.is_available() else nullcontext()
    gen_kwargs = dict(
        max_new_tokens=max_new_tokens,
        do_sample=do_sample,
        repetition_penalty=1.10,      # 같은 표현이 반복되는 것을 억제합니다.
        no_repeat_ngram_size=3,       # 3-gram 이상 동일 구절 반복을 금지합니다.
        eos_token_id=tokenizer.eos_token_id,
        pad_token_id=tokenizer.pad_token_id,
    )
    if do_sample:
        gen_kwargs.update(temperature=0.7, top_p=0.8)

    with torch.no_grad(), autocast_ctx:
        output = model.generate(**inputs, **gen_kwargs)
    # 입력 prompt 길이만큼 잘라내어, 새로 생성된 부분만 디코딩합니다.
    decoded = tokenizer.decode(output[0][inputs["input_ids"].shape[1]:], skip_special_tokens=False)
    return strip_qwen_think(decoded)


def count_trainable_params(m):
    '''전체 파라미터 수와, 그중 실제로 학습되는(requires_grad=True) 파라미터 수를 반환합니다.'''
    trainable = sum(p.numel() for p in m.parameters() if p.requires_grad)
    total = sum(p.numel() for p in m.parameters())
    return trainable, total


def count_lora_params(m):
    '''파라미터 이름에 "lora"가 포함된 것만 따로 집계합니다 (LoRA adapter 파라미터 전용 카운트).'''
    lora_total = 0
    lora_trainable = 0
    for name, p in m.named_parameters():
        if "lora" in name.lower():
            lora_total += p.numel()
            if p.requires_grad:
                lora_trainable += p.numel()
    return lora_trainable, lora_total


def ensure_lora_trainable(m, adapter_name: str = "default"):
    '''
    get_peft_model() 이후에도 LoRA 파라미터가 requires_grad=False로 남는 경우가 있어,
    여기서 명시적으로 다시 학습 가능하게 만들고, 그래도 0개면 바로 에러를 내서 알려줍니다.
    '''
    if not hasattr(m, "peft_config"):
        raise RuntimeError("현재 model은 PEFT/LoRA 모델이 아닙니다. get_peft_model() 이후에 호출하세요.")

    # PEFT 버전에 따라 set_adapter가 adapter를 trainable로 되살립니다.
    if hasattr(m, "set_adapter"):
        m.set_adapter(adapter_name)

    # 그래도 0개가 되는 경우를 막기 위해 LoRA 파라미터는 직접 requires_grad=True로 확인합니다.
    for name, p in m.named_parameters():
        if "lora" in name.lower():
            p.requires_grad_(True)

    m.train()
    m.config.use_cache = False

    trainable, total = count_trainable_params(m)
    lora_trainable, lora_total = count_lora_params(m)

    print(f"LoRA params: {lora_total:,}")
    print(f"Trainable LoRA params: {lora_trainable:,}")
    print(f"trainable params: {trainable:,} / {total:,} ({100 * trainable / total:.4f}%)")

    if lora_total == 0:
        raise RuntimeError("LoRA 파라미터가 없습니다. target_modules 또는 get_peft_model() 실행 여부를 확인하세요.")
    if lora_trainable == 0 or trainable == 0:
        raise RuntimeError("LoRA adapter가 붙었지만 trainable이 0개입니다. LoRA 설정 셀부터 다시 실행하세요.")
    return m
```

## 4. Before 응답 확인

학습 전 모델의 기본 응답을 저장합니다.


```python
test_questions = [
    "오늘 컨디션 어때?",
    "밥 먹었어?",
    "고마워요!",
    "요즘 스트레스가 심해",
    "주말에 뭐하지?",
]

# 학습 후 After 응답과 비교하기 위해, 파인튜닝 전(Before) 응답을 미리 저장해 둡니다.
before_answers = {}
print("=" * 60)
print("파인튜닝 전(Before) 응답")
print("=" * 60)
for q in test_questions:
    ans = ask(q, max_new_tokens=100, do_sample=False)
    before_answers[q] = ans
    print(f"\nQ: {q}\nA: {ans}")
    print("-" * 50)
```

```python
from datasets import Dataset

# instruction(질문)/response(사극체 답변) 쌍을 학습에 필요한 원본 형태로 변환합니다.
raw_data = [{"instruction": q, "response": a} for q, a in sageuk_pairs]

def format_example(example):
    # chat template을 적용해, 모델이 실제로 보게 될 형태의 prompt(질문)를 만듭니다.
    prompt = apply_chat(
        [{"role": "user", "content": example["instruction"]}],
        add_generation_prompt=True,
    )
    # completion에는 EOS를 직접 붙이지 않습니다.
    # SFTTrainer가 eos_token을 기준으로 필요한 EOS를 추가합니다.
    completion = example["response"].strip()
    return {"prompt": prompt, "completion": completion}

full_dataset = Dataset.from_list(raw_data).map(format_example)
# SFTTrainer가 가장 안정적으로 해석하도록 prompt/completion 두 컬럼만 남깁니다.
train_dataset = full_dataset.select_columns(["prompt", "completion"])

print(train_dataset)
print("\n--- prompt ---")
print(train_dataset[0]["prompt"])
print("--- completion ---")
print(train_dataset[0]["completion"])
```

## 6. LoRA 설정

중요한 순서입니다.

1. `prepare_model_for_kbit_training(model)`
2. `get_peft_model(model, lora_config)`
3. `ensure_lora_trainable(model)`
4. 그 다음에 `SFTTrainer(...)`

`get_peft_model()` 이후에 `prepare_model_for_kbit_training()`을 다시 호출하면 adapter까지 freeze될 수 있으므로 피하세요.


```python
from peft import LoraConfig, prepare_model_for_kbit_training, get_peft_model

if hasattr(model, "peft_config"):
    raise RuntimeError(
        "이미 LoRA adapter가 붙은 모델입니다. 이 셀을 중복 실행하지 말고, 모델 로드 셀부터 다시 실행하세요."
    )

# 4bit 양자화 모델에서도 안정적으로 backward가 가능하도록 준비합니다
# (일부 레이어를 fp32로 캐스팅하고, gradient checkpointing 설정을 켭니다).
model = prepare_model_for_kbit_training(model, use_gradient_checkpointing=True)
model.config.use_cache = False

lora_config = LoraConfig(
    r=16,                  # LoRA rank: adapter 행렬의 차원 (클수록 표현력↑, 파라미터/VRAM 사용량↑)
    lora_alpha=32,          # LoRA 스케일링 factor (보통 r의 2배 정도로 설정)
    lora_dropout=0.05,      # adapter에 적용하는 dropout, 과적합 방지용
    bias="none",            # bias 파라미터는 학습하지 않습니다.
    task_type="CAUSAL_LM",
    inference_mode=False,   # 학습 모드로 사용합니다 (추론 전용이면 True).
    target_modules=[        # LoRA adapter를 붙일 linear 레이어들 (attention + MLP projection)
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj",
    ],
    # 모델이 달라져 target_modules 에러가 나면 아래 한 줄로 대체할 수 있습니다.
    # target_modules="all-linear",
)

# base model에 LoRA adapter를 붙입니다. 이후 학습에서는 adapter 파라미터만 업데이트됩니다.
model = get_peft_model(model, lora_config)
model = ensure_lora_trainable(model, adapter_name="default")
```

## 7. 학습 설정 및 Trainer 생성

40개 데이터셋에서는 step별 loss가 상당히 출렁입니다. 그래서 `logging_steps=1`로 흐름을 보되, 판단은 최소 수십 step 이후에 합니다.


```python
from trl import SFTTrainer, SFTConfig

training_args = SFTConfig(
    output_dir="./outputs/sample_a_sageuk_qlora_improved",

    # 작은 데이터셋에서는 epoch보다 max_steps가 loss 변화를 보기 쉽습니다.
    max_steps=80,
    per_device_train_batch_size=1,      # GPU 1개당 한 번에 처리하는 샘플 수 (VRAM이 작아 1로 설정)
    gradient_accumulation_steps=8,      # 8 step만큼 gradient를 모아 업데이트 → 실질적인 배치 크기는 1*8=8

    gradient_checkpointing=True,        # activation을 저장하지 않고 필요할 때 재계산해 VRAM을 아낍니다.
    gradient_checkpointing_kwargs={"use_reentrant": False},

    learning_rate=2e-4,
    warmup_ratio=0.05,      # 전체 step의 5%까지 learning rate를 서서히 올립니다.
    max_grad_norm=0.3,      # gradient clipping: 너무 큰 gradient로 학습이 튀는 것을 방지합니다.
    lr_scheduler_type="cosine",  # learning rate를 cosine 곡선을 따라 서서히 줄여나갑니다.

    logging_steps=1,        # 매 step마다 loss를 기록해, 작은 데이터셋의 변화를 세밀히 관찰합니다.
    save_strategy="no",     # 실습용이라 중간 체크포인트는 저장하지 않습니다.

    bf16=BF16_OK,
    fp16=not BF16_OK,          # BF16을 지원하지 않는 GPU에서는 FP16으로 대체합니다.
    optim="paged_adamw_8bit",  # 8bit AdamW + paging으로 optimizer state의 VRAM 사용량을 줄입니다.
    max_length=512,             # 이보다 긴 시퀀스는 잘라냅니다.

    # prompt-completion dataset에서는 completion 부분만 loss를 주는 것이 목적에 맞습니다.
    completion_only_loss=True,
    eos_token=EOS_TOKEN,

    report_to="wandb",              # ← "none" 대신 "wandb"로 변경
    run_name="qlora-sageuk-v1",     # (선택) wandb 대시보드에 표시될 실행 이름
)

trainer = SFTTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    processing_class=tokenizer,
)

# Trainer 생성 이후에도 trainable이 살아있는지 재확인합니다.
trainer.model = ensure_lora_trainable(trainer.model, adapter_name="default")
```

## 8. Loss label masking 진단

이 셀에서 decoded label에 질문 문장이나 `<|im_start|>user`가 많이 보이면 prompt까지 loss가 걸린 것입니다. 정상이라면 사극체 답변 부분 위주로 보입니다.


```python
# SFTTrainer가 만든 첫 학습 배치를 하나 뽑아서, loss가 실제로 어느 토큰에 걸리는지 확인합니다.
batch = next(iter(trainer.get_train_dataloader()))
print(batch.keys())
print("input_ids shape:", tuple(batch["input_ids"].shape))
print("labels shape   :", tuple(batch["labels"].shape))

# labels 중 -100은 loss 계산에서 제외되는 토큰입니다 (주로 prompt 부분).
# -100이 아닌 토큰 수 = 실제로 loss가 걸리는 토큰 수입니다.
loss_token_count = int((batch["labels"] != -100).sum().item())
print("loss가 걸리는 token 수:", loss_token_count)

for i in range(min(2, batch["labels"].shape[0])):
    # -100을 제외한 label 토큰만 디코딩하면, loss가 걸리는 부분(completion)만 보입니다.
    label_ids = batch["labels"][i][batch["labels"][i] != -100]
    print("\n---- decoded labels ----")
    print(tokenizer.decode(label_ids, skip_special_tokens=False))
```

## 9. 학습 실행

`trainable params`가 0이 아닌 것을 확인한 뒤 실행합니다. 학습 직전부터 최대 VRAM을 다시 측정합니다.


```python
# 학습 시작 전 GPU 캐시와 피크 메모리 통계를 초기화해, 이번 학습에서 쓴 VRAM만 측정합니다.
clear_cuda_cache()
if torch.cuda.is_available():
    torch.cuda.reset_peak_memory_stats()

# adapter가 trainable 상태인지 학습 실행 직전에 한 번 더 확인합니다.
trainer.model = ensure_lora_trainable(trainer.model, adapter_name="default")


from transformers import TrainerCallback
import time

# 기본 로그는 step마다 dict로만 출력되어 흐름을 보기 어려우므로,
# 진행률(%)과 예상 잔여 시간(ETA)을 함께 보여주는 콜백을 추가합니다.
class ProgressPrintCallback(TrainerCallback):
    def on_train_begin(self, args, state, control, **kwargs):
        self._start_time = time.time()

    def on_log(self, args, state, control, logs=None, **kwargs):
        if logs is None or "loss" not in logs:
            return
        elapsed = time.time() - self._start_time
        pct = state.global_step / state.max_steps * 100 if state.max_steps else 0
        eta = (elapsed / state.global_step * (state.max_steps - state.global_step)) if state.global_step else 0
        print(
            f"[{state.global_step}/{state.max_steps}] {pct:5.1f}% | "
            f"loss={logs['loss']:.4f} | lr={logs.get('learning_rate', 0):.2e} | "
            f"경과 {elapsed:6.1f}s | 예상 잔여 {eta:6.1f}s"
        )

trainer.add_callback(ProgressPrintCallback())

# 실제 학습 루프 실행 (SFTConfig에서 정한 max_steps만큼 진행합니다)
train_result = trainer.train()

model.eval()  # 학습이 끝났으니 추론 모드로 전환합니다.
print("학습 완료! model.eval() 적용됨")

print("=" * 60)
print("학습 리소스 요약")
print("=" * 60)
print(f"학습 소요 시간: {train_result.metrics.get('train_runtime', float('nan')):.1f}초")
if torch.cuda.is_available():
    peak_vram = torch.cuda.max_memory_allocated() / 1024**3
    print(f"학습 중 최대 VRAM 사용량: {peak_vram:.2f} GB")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# trainer.state.log_history에는 매 logging_steps마다 기록된 loss가 쌓여 있습니다.
# 그중 loss와 step 정보가 있는 항목만 추출합니다.
loss_rows = [
    {"step": row["step"], "loss": row["loss"]}
    for row in trainer.state.log_history
    if "loss" in row and "step" in row
]

loss_df = pd.DataFrame(loss_rows)
display(loss_df.tail(20))  # 마지막 20개 step만 표로 확인합니다.

if len(loss_df) > 0:
    # step에 따른 loss 변화 추세를 그래프로 확인합니다 (개별 step보다 전체 추세가 중요합니다).
    ax = loss_df.plot(x="step", y="loss", marker="o", title="Training loss")
    ax.set_xlabel("optimizer step")
    ax.set_ylabel("loss")
    plt.show()
```

```python
import os
import warnings

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"
warnings.filterwarnings("ignore")

import tensorflow as tf
print(tf.__version__)
```

```python
print("GPU:", tf.config.list_physical_devices("GPU"))
```
